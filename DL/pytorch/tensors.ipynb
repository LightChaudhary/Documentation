```python
random_seed = 0
torch.manual_seed(random_seed)
random_tensor_A = torch.rand(size=(7,7))
torch.manual_seed(random_seed)
random_tensor_B = torch.rand(size=(1,7))

torch.mm(random_tensor_A,random_tensor_B.T), random_tensor_A, random_tensor_B
```

```python
%%time
device = torch.device("mps") if torch.backends.mps.is_available() else torch.device("cpu")
random_seed = 1234
torch.manual_seed(seed=random_seed)
tensor_A = torch.rand(size=(7,7), device=device)
torch.manual_seed(seed=random_seed)
tensor_B = torch.rand(size=(1,7), device=device)

result = torch.matmul(tensor_A, tensor_B.T)

print(result.to("cpu"))
```
