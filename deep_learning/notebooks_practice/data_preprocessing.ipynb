```python
import os

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file= os.path.join('..', 'data', 'student.csv')

with open(data_file, 'w') as file:
    file.write('''Age,Gender,StudyHours,Score
20,Male,3,72
NA,Female,5,85
22,NA,4,78
21,Male,NA,91
19,Female,2,65''')
```

```python
import pandas as pd

df = pd.read_csv(data_file)
df, df.shape, df.dtypes
```

```python
inputs, target = df.iloc[:, :3], df.iloc[:, 3]
inputs, target
```

```python
inputs = pd.get_dummies(inputs, dummy_na=True)
inputs, inputs.dtypes
```
