```python
X = torch.arange(12, dtype=torch.float32).reshape(3, 4)

Y = torch.tensor([
    [2.0, 1, 4, 3],
    [1, 2, 3, 4],
    [4, 3, 2, 1]
])

print(torch.cat((X,Y),dim=0)) # (6,4)
print(torch.cat((X,Y),dim=1)) # (3,8)
```
