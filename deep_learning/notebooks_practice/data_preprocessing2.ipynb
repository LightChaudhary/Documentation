```python
import os
import pandas as pd
import torch

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'housing.csv')

with open(data_file, 'w') as f:
    f.write('''City,Type,Bedrooms,Price
Kathmandu,Apartment,2,85000
Pokhara,House,3,120000
NA,Apartment,2,90000
Lalitpur,NA,4,150000
Kathmandu,House,NA,110000
Pokhara,Apartment,1,70000''')

df = pd.read_csv(data_file)

inputs, target = df.iloc[:, :3], df.iloc[:, 3]

inputs = pd.get_dummies(inputs, dummy_na=True)
inputs= inputs.fillna(inputs.mean())

X = torch.tensor(inputs.to_numpy(dtype=float))
y = torch.tensor(target.to_numpy(dtype=float))
```
