# Linear Regression

##### Learn some basic prediction equation, cost function (MSE) equation, normal equation for value θ of for a Linear Regression model from hands on machine learning with scikit-learn page no 112-113

```python
import numpy as np

X = 2 * np.random.rand(100,1) # random.rand(rows, columns) generate random numbers between 0 to 1.
y = 4 + 3 * X + np.random.randn(100, 1) # np.random.randn(rows, columns) mostly used for noise.
```

```python
X_b = np.c_[np.ones((100,1)), X] # np.ones((100,1)) It creates a column of 100 ones. np.c_[] c_ means column concatenate. 
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y) # It is exactly this mathematical formula: (θ=(XᵀX)⁻¹.Xᵀy) This is called the Normal Equation.
```

```python
X_new = np.array([[0], [2]])
X_new_b = np.c_[np.ones((2,1)), X_new] 
y_predict = X_new_b.dot(theta_best)
y_predict
```

```python
import matplotlib.pyplot as plt
plt.plot(X_new, y_predict, "r-")
plt.plot(X,y,"b.")
plt.axis([0,2,0,15])
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
lin_reg.intercept_, lin_reg.coef_
```

```python
lin_reg.predict(X_new)
```
